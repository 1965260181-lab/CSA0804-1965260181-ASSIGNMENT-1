```python
BASE_RATE = 30.00
FAT_STEP_VALUE = 0.40
SNF_STEP_VALUE = 0.30

STANDARD_FAT = 3.5
STANDARD_SNF = 8.5
STEP_SIZE = 0.1

litres = 10
fat = 4.2
snf = 8.6

fat_steps = round((fat - STANDARD_FAT) / STEP_SIZE)
snf_steps = round((snf - STANDARD_SNF) / STEP_SIZE)

rate = BASE_RATE + (fat_steps * FAT_STEP_VALUE) + (snf_steps * SNF_STEP_VALUE)

payment = round(litres * rate, 2)

print("Rate:", rate)
print("Payment:", payment)
```
