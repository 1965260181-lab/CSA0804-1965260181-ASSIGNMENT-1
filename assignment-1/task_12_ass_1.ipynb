```python
%%writefile milkrate_v2.py

from decimal import Decimal, ROUND_HALF_UP

NORMAL_CARD = {
    "base": Decimal("30.00"),
    "fat_step": Decimal("0.40"),
    "snf_step": Decimal("0.30")
}

SUMMER_CARD = {
    "base": Decimal("32.00"),
    "fat_step": Decimal("0.45"),
    "snf_step": Decimal("0.30")
}

def select_rate_card(month):
    return SUMMER_CARD if month in (4, 5, 6) else NORMAL_CARD

def rate_per_litre(fat, snf, card):
    fat_steps = (fat - Decimal("3.5")) / Decimal("0.1")
    snf_steps = (snf - Decimal("8.5")) / Decimal("0.1")

    rate = card["base"] + (fat_steps * card["fat_step"]) + (snf_steps * card["snf_step"])

    return rate.quantize(Decimal("0.01"), rounding=ROUND_HALF_UP)

def bonus(total_litres):
    if total_litres > Decimal("500"):
        return (total_litres * Decimal("0.50")).quantize(
            Decimal("0.01"), rounding=ROUND_HALF_UP
        )
    return Decimal("0.00")
```
