```python
def read_float(prompt):
    while True:
        text = input(prompt)

        try:
            val = float(text)

            if val < 0:
                print("Value cannot be negative.")
                continue

            return val

        except ValueError:
            print("Invalid input! Provide numeric digits (e.g. 8.5).")


value = read_float("Enter a number: ")
print(value)
```
