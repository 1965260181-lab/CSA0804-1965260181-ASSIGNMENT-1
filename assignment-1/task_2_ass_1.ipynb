```python
from decimal import Decimal

# 1. Type inspection
print(type(10))
print(type(10.0))
print(type("10"))
print(type(True))
print(type([10, 8.5]))

# 2. Division and Floor Arithmetic
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(-7 // 2)

# 3. Floating-point precision
print(0.1 + 0.2 == 0.3)

print(Decimal('0.1') + Decimal('0.2') == Decimal('0.3'))
```
