```python
BASE_RATE = 30.00
FAT_STEP_VALUE = 0.40
SNF_STEP_VALUE = 0.30

STANDARD_FAT = 3.5
STANDARD_SNF = 8.5
STEP_SIZE = 0.1


def rate_per_litre(fat, snf):
    """Calculate rate based on standard steps."""

    fat_steps = round((fat - STANDARD_FAT) / STEP_SIZE)
    snf_steps = round((snf - STANDARD_SNF) / STEP_SIZE)

    return round(BASE_RATE + (fat_steps * FAT_STEP_VALUE) + (snf_steps * SNF_STEP_VALUE), 2)


def payout(litres, fat, snf):
    """Calculate payment amount for given litres."""

    return round(litres * rate_per_litre(fat, snf), 2)


def ten_day_total(entries):
    """Sum total payouts over 10 days."""

    return round(sum(payout(l, f, s) for l, f, s in entries), 2)


print("Rate:", rate_per_litre(3.5, 8.5))
print("Payment:", payout(10, 3.5, 8.5))

entries = [
    (10, 3.5, 8.5),
    (12, 4.0, 9.0)
]

print("Ten Day Total:", ten_day_total(entries))
```
